# FabricFit Ghana — conservative 8-class baseline

This notebook runs only on a reviewed, licensed dataset with group-disjoint splits. It does not scrape or infer labels in Kaggle. GPU is used when the Kaggle session provides one; CPU otherwise. Test metrics are produced only after the trainer's minimum-data gate passes.

```python
from pathlib import Path
import json, os, subprocess, sys
print('TensorFlow:', __import__('tensorflow').__version__)
print('GPUs:', __import__('tensorflow').config.list_physical_devices('GPU'))
print('Input mounts:', [p.name for p in Path('/kaggle/input').iterdir()])
DATA = next((p for p in Path('/kaggle/input').rglob('processed/labels.csv')), None)
if DATA is None: raise RuntimeError('Attach a prepared FabricFit dataset containing dataset/processed/labels.csv')
DATA = DATA.parent
PROJECT = next(p for p in Path('/kaggle/input').rglob('scripts/train_model.py')) .parent.parent
OUT = Path('/kaggle/working/fabricfit_output'); OUT.mkdir(exist_ok=True)
print('Dataset:', DATA, 'Project scripts:', PROJECT, 'Output:', OUT)
```

```python
cmd = [sys.executable, str(PROJECT/'scripts/train_model.py'), '--data', str(DATA), '--out', str(OUT/'models'), '--reports', str(OUT/'reports')]
print('Launching two-stage MobileNetV3Small. The script fails closed below the 300-per-class training bar.')
subprocess.run(cmd, check=True)
```

```python
for p in sorted(OUT.rglob('*')):
    if p.is_file(): print(p.relative_to(OUT), p.stat().st_size, 'bytes')
print('Metrics:', (OUT/'reports/model_comparison.json').read_text())
```
